```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../../data/Ames_final_df_final.csv')
```

```python
df.head()
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

```python
df.info()
```

```python
X = df.drop('SalePrice', axis=1)
y = df['SalePrice']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=51)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
```

```python
X_train_scaled = scaler.transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import ElasticNet
```

```python
model = ElasticNet(max_iter=1000000)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_param = {'alpha':[0.1, 1, 5, 10, 50, 100], 'l1_ratio':[0.1, 0.3, 0.5, 0.8, 0.95, 0.99, 1]}
```

```python
grid_model = GridSearchCV(estimator=model, 
                          param_grid=grid_param,
                          scoring='neg_mean_squared_error',
                          cv=10,
                          verbose=1)
```

```python
grid_model.fit(X_train_scaled, y_train)
```

```python
grid_model.best_estimator_
```

```python
results = pd.DataFrame(grid_model.cv_results_)
```

```python
results.head()
```

```python
y_predicitons = grid_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
print(f"mae: {mean_absolute_error(y_test, y_predicitons)}")
print(f"mse: {np.sqrt(mean_squared_error(y_test, y_predicitons))}")
```
